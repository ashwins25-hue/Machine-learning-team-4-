```python
# ============================================================
# SAFEHABITAT AI
# Chennai Multi-Hazard Disaster Risk Prediction,
# Red-Zone Identification and Relocation Priority Assessment
# ============================================================
#
# MICRO-PROJECT
#
# TECHNIQUES:
# 1. Data Preprocessing
# 2. Feature Selection
# 3. Random Forest Classification
# 4. Disaster Risk Prediction
# 5. Red-Zone Identification
# 6. Vulnerability Assessment
# 7. Relocation Priority Calculation
# 8. Candidate Site Suitability Assessment
# 9. Data Visualization
# 10. Export Results
#
# DATASETS REQUIRED:
# safehabitat_chennai_multihazard_synthetic.csv
# safehabitat_chennai_candidate_sites_synthetic.csv
#
# NOTE:
# Both datasets are synthetic demonstration datasets.
# Predictions and site assessments are NOT official
# disaster-risk or relocation recommendations.
# ============================================================


# ============================================================
# STEP 1: IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
from IPython.display import display

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

import joblib

sns.set_theme(style="whitegrid")

print("=" * 60)
print("          SAFEHABITAT AI")
print("     CHENNAI MULTI-HAZARD RISK ASSESSMENT")
print("=" * 60)


# ============================================================
# STEP 2: UPLOAD BOTH DATASETS
# ============================================================

print("\nUPLOAD BOTH CSV DATASETS:")
print("1. safehabitat_chennai_multihazard_synthetic.csv")
print("2. safehabitat_chennai_candidate_sites_synthetic.csv")

uploaded = files.upload()

main_file = "safehabitat_chennai_multihazard_synthetic.csv"
site_file = "safehabitat_chennai_candidate_sites_synthetic.csv"

if main_file not in uploaded:
    raise FileNotFoundError(
        "Main Chennai dataset was not uploaded."
    )

if site_file not in uploaded:
    raise FileNotFoundError(
        "Candidate relocation dataset was not uploaded."
    )

df = pd.read_csv(main_file)
sites = pd.read_csv(site_file)

print("\nDATASETS LOADED SUCCESSFULLY")

print("\nMain Dataset Shape:", df.shape)
print("Relocation Sites Shape:", sites.shape)

print("\nMAIN DATASET SAMPLE:")
display(df.head())

print("\nRELOCATION SITES SAMPLE:")
display(sites.head())


# ============================================================
# STEP 3: DATASET INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("STEP 3: DATASET EXPLORATION")
print("=" * 60)

print("\nDATASET COLUMNS:")
print(df.columns.tolist())

print("\nDATASET INFORMATION:")
df.info()

print("\nMISSING VALUES:")
print(df.isnull().sum())

print("\nDUPLICATE RECORD IDs:")
print(df["record_id"].duplicated().sum())

print("\nRISK LEVEL DISTRIBUTION:")
print(df["risk_level"].value_counts())

print("\nHAZARD TYPE DISTRIBUTION:")
print(df["dominant_hazard"].value_counts())

print("\nSTATISTICAL SUMMARY:")
display(df.describe(include="all"))


# ============================================================
# STEP 4: DATA CLEANING
# ============================================================

print("\n" + "=" * 60)
print("STEP 4: DATA CLEANING")
print("=" * 60)

# Remove duplicate record IDs
df = df.drop_duplicates(subset=["record_id"]).copy()

# Environmental numerical features
numeric_features = [
    "rainfall_mm_7d",
    "max_wind_kmph",
    "max_temperature_c",
    "elevation_m",
    "slope_degrees",
    "soil_moisture_pct",
    "distance_to_river_km",
    "distance_to_coast_km",
    "water_scarcity_index",
    "previous_disaster_events"
]

# Categorical feature
categorical_features = [
    "soil_type"
]

# Combined feature list
feature_cols = numeric_features + categorical_features

# Risk target categories
risk_classes = [
    "Low",
    "Moderate",
    "High"
]

# Remove records without valid target labels
df = df[
    df["risk_level"].isin(risk_classes)
].copy()

# Convert numeric columns safely
for column in numeric_features:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

print("\nDATA CLEANING COMPLETED")
print("Cleaned Dataset Shape:", df.shape)


# ============================================================
# STEP 5: FEATURE SELECTION
# ============================================================

print("\n" + "=" * 60)
print("STEP 5: FEATURE SELECTION")
print("=" * 60)

X = df[feature_cols].copy()
y = df["risk_level"].copy()

print("\nSELECTED ENVIRONMENTAL FEATURES:")

for feature in feature_cols:
    print("-", feature)

print("\nTARGET VARIABLE: risk_level")

print("\nFEATURE MATRIX SHAPE:", X.shape)


# ============================================================
# STEP 6: TRAIN / TEST SPLIT
# ============================================================

print("\n" + "=" * 60)
print("STEP 6: TRAIN / TEST SPLIT")
print("=" * 60)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Records:", len(X_train))
print("Testing Records:", len(X_test))


# ============================================================
# STEP 7: DATA PREPROCESSING PIPELINE
# ============================================================

print("\n" + "=" * 60)
print("STEP 7: PREPROCESSING PIPELINE")
print("=" * 60)

# Handle missing numerical values
numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)

# Handle missing categorical values and encoding
categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(handle_unknown="ignore")
        )
    ]
)

# Combine numerical and categorical preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_transformer,
            numeric_features
        ),
        (
            "categorical",
            categorical_transformer,
            categorical_features
        )
    ]
)

print("\nPREPROCESSING PIPELINE CREATED")


# ============================================================
# STEP 8: RANDOM FOREST MODEL
# ============================================================

print("\n" + "=" * 60)
print("STEP 8: RANDOM FOREST MODEL TRAINING")
print("=" * 60)

rf_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=200,
                random_state=42,
                class_weight="balanced",
                n_jobs=-1
            )
        )
    ]
)

rf_model.fit(X_train, y_train)

print("\nRANDOM FOREST MODEL TRAINED SUCCESSFULLY")


# ============================================================
# STEP 9: MODEL PREDICTION
# ============================================================

print("\n" + "=" * 60)
print("STEP 9: MODEL PREDICTION")
print("=" * 60)

y_pred = rf_model.predict(X_test)

print("\nTEST SET PREDICTIONS:")
print(y_pred[:20])


# ============================================================
# STEP 10: MODEL EVALUATION
# ============================================================

print("\n" + "=" * 60)
print("STEP 10: MODEL EVALUATION")
print("=" * 60)

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

high_recall = recall_score(
    y_test,
    y_pred,
    labels=["High"],
    average="macro",
    zero_division=0
)

print("\nMODEL PERFORMANCE")
print("-" * 40)

print("Accuracy :", round(accuracy * 100, 2), "%")
print("Precision:", round(precision * 100, 2), "%")
print("Recall   :", round(recall * 100, 2), "%")
print("F1 Score :", round(f1 * 100, 2), "%")
print("High-Risk Recall:", round(high_recall * 100, 2), "%")

print("\nCLASSIFICATION REPORT:")

print(
    classification_report(
        y_test,
        y_pred,
        labels=risk_classes,
        zero_division=0
    )
)


# ============================================================
# STEP 11: CONFUSION MATRIX
# ============================================================

print("\n" + "=" * 60)
print("STEP 11: CONFUSION MATRIX")
print("=" * 60)

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=risk_classes
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Greys",
    xticklabels=risk_classes,
    yticklabels=risk_classes
)

plt.xlabel("Predicted Risk Level")
plt.ylabel("Actual Risk Level")
plt.title("SafeHabitat AI - Confusion Matrix")
plt.tight_layout()
plt.show()


# ============================================================
# STEP 12: FEATURE IMPORTANCE
# ============================================================

print("\n" + "=" * 60)
print("STEP 12: FEATURE IMPORTANCE")
print("=" * 60)

forest = rf_model.named_steps["classifier"]

feature_names = (
    rf_model.named_steps["preprocessor"]
    .get_feature_names_out()
)

importance_df = pd.DataFrame(
    {
        "Feature": feature_names,
        "Importance": forest.feature_importances_
    }
)

importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

print("\nFEATURE IMPORTANCE TABLE:")
display(importance_df)

plt.figure(figsize=(10, 6))

plt.barh(
    importance_df["Feature"],
    importance_df["Importance"],
    color="gray"
)

plt.xlabel("Feature Importance")
plt.ylabel("Environmental Features")
plt.title("Random Forest Feature Importance")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()


# ============================================================
# STEP 13: PREDICT RISK FOR ALL HABITATIONS
# ============================================================

print("\n" + "=" * 60)
print("STEP 13: DISASTER RISK PREDICTION")
print("=" * 60)

# Predict class
df["predicted_risk"] = rf_model.predict(
    df[feature_cols]
)

# Predict class probabilities
probabilities = rf_model.predict_proba(
    df[feature_cols]
)

# Identify class index for High
class_names = list(rf_model.classes_)

high_index = class_names.index("High")

# Probability of High class
df["high_risk_score"] = probabilities[:, high_index]

# Model confidence percentage
df["prediction_confidence"] = (
    probabilities.max(axis=1) * 100
)

print("\nPREDICTED DISASTER RISK:")

display(
    df[
        [
            "record_id",
            "locality",
            "dominant_hazard",
            "predicted_risk",
            "high_risk_score",
            "prediction_confidence"
        ]
    ].head(20)
)


# ============================================================
# STEP 14: RED-ZONE IDENTIFICATION
# ============================================================

print("\n" + "=" * 60)
print("STEP 14: RED-ZONE IDENTIFICATION")
print("=" * 60)

# High-risk records become potential red zones
df["potential_red_zone"] = (
    df["predicted_risk"] == "High"
)

red_zones = df[
    df["potential_red_zone"]
].copy()

print("\nTOTAL POTENTIAL RED-ZONE RECORDS:", len(red_zones))

print("\nRED-ZONE HABITATIONS:")

display(
    red_zones[
        [
            "record_id",
            "locality",
            "dominant_hazard",
            "predicted_risk",
            "high_risk_score"
        ]
    ].head(20)
)


# ============================================================
# STEP 15: VULNERABILITY ASSESSMENT
# ============================================================

print("\n" + "=" * 60)
print("STEP 15: VULNERABILITY ASSESSMENT")
print("=" * 60)

# Normalize numerical values between 0 and 1
def normalize_01(series):

    series = pd.to_numeric(
        series,
        errors="coerce"
    )

    minimum = series.min()
    maximum = series.max()

    if (
        pd.isna(minimum)
        or pd.isna(maximum)
        or maximum == minimum
    ):
        return pd.Series(
            0.0,
            index=series.index
        )

    normalized = (
        (series - minimum) /
        (maximum - minimum)
    )

    return normalized.fillna(0)


# Population vulnerability
df["population_vulnerability"] = normalize_01(
    df["population"]
)

# Housing vulnerability
housing_mapping = {
    "Good": 0.0,
    "Average": 0.5,
    "Poor": 1.0
}

df["housing_vulnerability"] = (
    df["housing_condition"]
    .map(housing_mapping)
    .fillna(0.5)
)

# Road accessibility
road_mapping = {
    "Good": 0.0,
    "Moderate": 0.5,
    "Poor": 1.0
}

road_score = (
    df["road_access"]
    .map(road_mapping)
    .fillna(0.5)
)

# Healthcare accessibility
healthcare_score = normalize_01(
    df["healthcare_access_km"]
)

# Combined accessibility vulnerability
df["accessibility_vulnerability"] = (
    0.50 * road_score +
    0.50 * healthcare_score
)

# Historical disaster exposure
df["exposure_vulnerability"] = normalize_01(
    df["previous_disaster_events"]
)

print("\nVULNERABILITY ASSESSMENT COMPLETED")

display(
    df[
        [
            "locality",
            "population_vulnerability",
            "housing_vulnerability",
            "accessibility_vulnerability",
            "exposure_vulnerability"
        ]
    ].head(20)
)


# ============================================================
# STEP 16: RELOCATION PRIORITY SCORE
# ============================================================

print("\n" + "=" * 60)
print("STEP 16: RELOCATION PRIORITY CALCULATION")
print("=" * 60)

# Illustrative weights (total = 1.0)
WEIGHT_RISK = 0.40
WEIGHT_POPULATION = 0.20
WEIGHT_HOUSING = 0.15
WEIGHT_ACCESSIBILITY = 0.10
WEIGHT_EXPOSURE = 0.15

df["relocation_priority_score"] = 100 * (
    WEIGHT_RISK * df["high_risk_score"] +
    WEIGHT_POPULATION * df["population_vulnerability"] +
    WEIGHT_HOUSING * df["housing_vulnerability"] +
    WEIGHT_ACCESSIBILITY * df["accessibility_vulnerability"] +
    WEIGHT_EXPOSURE * df["exposure_vulnerability"]
)


# Priority category
def get_priority(score):

    if score >= 70:
        return "High Priority"

    elif score >= 40:
        return "Medium Priority"

    else:
        return "Low Priority"


df["priority_category"] = (
    df["relocation_priority_score"]
    .apply(get_priority)
)

# Sort only predicted high-risk records
priority_habitations = (
    df[df["potential_red_zone"]]
    .sort_values(
        by="relocation_priority_score",
        ascending=False
    )
    .copy()
)

print("\nRELOCATION PRIORITY RANKING:")

display(
    priority_habitations[
        [
            "record_id",
            "locality",
            "population",
            "predicted_risk",
            "relocation_priority_score",
            "priority_category"
        ]
    ].head(20)
)


# ============================================================
# STEP 17: CANDIDATE RELOCATION SITE ASSESSMENT
# ============================================================

print("\n" + "=" * 60)
print("STEP 17: CANDIDATE RELOCATION SITE ASSESSMENT")
print("=" * 60)

if priority_habitations.empty:

    print("No high-risk habitation records found.")

    sites["fully_eligible"] = False

else:

    # Select the highest-ranked high-risk record
    selected_habitation = priority_habitations.iloc[0]

    selected_locality = selected_habitation["locality"]

    required_population = int(
        selected_habitation["population"]
    )

    print("\nSELECTED HABITATION:", selected_locality)
    print("POPULATION:", required_population)

    # Convert site capacity to numerical
    sites["available_capacity_persons"] = pd.to_numeric(
        sites["available_capacity_persons"],
        errors="coerce"
    )

    # Capacity requirement
    sites["capacity_ok"] = (
        sites["available_capacity_persons"] >= required_population
    )

    # Water availability
    sites["water_ok"] = (
        pd.to_numeric(
            sites["water_available"],
            errors="coerce"
        ) == 1
    )

    # Road accessibility
    sites["road_ok"] = (
        sites["road_access"] == "Good"
    )

    # Healthcare accessibility (illustrative threshold)
    sites["healthcare_ok"] = (
        pd.to_numeric(
            sites["healthcare_distance_km"],
            errors="coerce"
        ) <= 5
    )

    # Independent hazard screening
    sites["safety_ok"] = (
        pd.to_numeric(
            sites["independent_hazard_screening_passed"],
            errors="coerce"
        ) == 1
    )

    # Land-use approval
    sites["approval_ok"] = (
        pd.to_numeric(
            sites["land_use_approval_confirmed"],
            errors="coerce"
        ) == 1
    )

    # Preliminary logistics suitability
    sites["logistics_ok"] = sites[
        [
            "capacity_ok",
            "water_ok",
            "road_ok",
            "healthcare_ok"
        ]
    ].all(axis=1)

    # Complete eligibility
    sites["fully_eligible"] = sites[
        [
            "capacity_ok",
            "water_ok",
            "road_ok",
            "healthcare_ok",
            "safety_ok",
            "approval_ok"
        ]
    ].all(axis=1)

    # Human-readable result
    sites["site_status"] = np.where(
        sites["fully_eligible"],
        "All prototype checks passed",
        "Not fully eligible / verification required"
    )

    print("\nSITE SUITABILITY RESULTS:")

    display(
        sites[
            [
                "site_name",
                "available_capacity_persons",
                "capacity_ok",
                "water_ok",
                "road_ok",
                "healthcare_ok",
                "safety_ok",
                "approval_ok",
                "fully_eligible",
                "site_status"
            ]
        ]
    )

    print(
        "\nSITES PASSING ALL CHECKS:",
        sites["fully_eligible"].sum()
    )


# ============================================================
# STEP 18: RISK DISTRIBUTION VISUALIZATION
# ============================================================

print("\n" + "=" * 60)
print("STEP 18: RISK DISTRIBUTION")
print("=" * 60)

risk_counts = (
    df["predicted_risk"]
    .value_counts()
    .reindex(
        risk_classes,
        fill_value=0
    )
)

plt.figure(figsize=(8, 5))

plt.bar(
    risk_counts.index,
    risk_counts.values,
    color=["lightgray", "gray", "black"]
)

plt.xlabel("Disaster Risk Level")
plt.ylabel("Number of Scenario Records")
plt.title("Chennai Multi-Hazard Risk Classification")
plt.tight_layout()
plt.show()


# ============================================================
# STEP 19: HAZARD DISTRIBUTION VISUALIZATION
# ============================================================

print("\n" + "=" * 60)
print("STEP 19: HAZARD DISTRIBUTION")
print("=" * 60)

hazard_counts = df["dominant_hazard"].value_counts()

plt.figure(figsize=(10, 6))

plt.barh(
    hazard_counts.index,
    hazard_counts.values,
    color="gray"
)

plt.xlabel("Number of Records")
plt.ylabel("Dominant Hazard")
plt.title("Multi-Hazard Distribution in Synthetic Chennai Data")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()


# ============================================================
# STEP 20: RELOCATION PRIORITY VISUALIZATION
# ============================================================

print("\n" + "=" * 60)
print("STEP 20: RELOCATION PRIORITY VISUALIZATION")
print("=" * 60)

top_priority = priority_habitations.head(10)

if not top_priority.empty:

    plt.figure(figsize=(10, 6))

    labels = (
        top_priority["locality"].astype(str)
        + " ("
        + top_priority["record_id"].astype(str)
        + ")"
    )

    plt.barh(
        labels,
        top_priority["relocation_priority_score"],
        color="gray"
    )

    plt.xlabel("Relocation Priority Score (%)")
    plt.ylabel("Habitation Scenario")
    plt.title("Top 10 Relocation Priority Records")
    plt.xlim(0, 100)
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()

else:

    print("No high-risk records for priority visualization.")


# ============================================================
# STEP 21: SAVE MODEL
# ============================================================

print("\n" + "=" * 60)
print("STEP 21: SAVE TRAINED MODEL")
print("=" * 60)

model_filename = "safehabitat_random_forest_model.pkl"

joblib.dump(
    rf_model,
    model_filename
)

print("Trained model saved:", model_filename)


# ============================================================
# STEP 22: EXPORT RESULTS
# ============================================================

print("\n" + "=" * 60)
print("STEP 22: EXPORT RESULTS")
print("=" * 60)

# Complete prediction dataset
df.to_csv(
    "safehabitat_chennai_predictions.csv",
    index=False
)

# Red-zone records
red_zones.to_csv(
    "safehabitat_red_zones.csv",
    index=False
)

# Relocation priority ranking
priority_habitations.to_csv(
    "safehabitat_relocation_priority.csv",
    index=False
)

# Relocation sites
sites.to_csv(
    "safehabitat_candidate_site_results.csv",
    index=False
)

# Model performance table
performance_df = pd.DataFrame(
    {
        "Metric": [
            "Accuracy",
            "Weighted Precision",
            "Weighted Recall",
            "Weighted F1 Score",
            "High Risk Recall"
        ],
        "Score": [
            accuracy,
            precision,
            recall,
            f1,
            high_recall
        ]
    }
)

performance_df.to_csv(
    "safehabitat_model_performance.csv",
    index=False
)

print("\nRESULT FILES SAVED SUCCESSFULLY")


# ============================================================
# STEP 23: FINAL PROJECT SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("       SAFEHABITAT AI - FINAL SUMMARY")
print("=" * 60)

print("\nTotal Scenario Records:", len(df))

print(
    "Potential Red-Zone Records:",
    len(red_zones)
)

print(
    "High Priority Records:",
    (df["priority_category"] == "High Priority").sum()
)

print(
    "Medium Priority Records:",
    (df["priority_category"] == "Medium Priority").sum()
)

print(
    "Low Priority Records:",
    (df["priority_category"] == "Low Priority").sum()
)

print("\nModel Accuracy:", round(accuracy * 100, 2), "%")
print("High-Risk Recall:", round(high_recall * 100, 2), "%")

if "fully_eligible" in sites.columns:

    print(
        "Candidate Sites Passing All Checks:",
        int(sites["fully_eligible"].sum())
    )

print("\nIMPORTANT:")
print("The input datasets contain synthetic demonstration data.")
print("Results are not verified Chennai hazard classifications.")
print("Site screening does not replace expert or government approval.")


# ============================================================
# STEP 24: DOWNLOAD RESULT FILES
# ============================================================

print("\nDOWNLOADING PROJECT OUTPUT FILES...")

output_files = [
    "safehabitat_chennai_predictions.csv",
    "safehabitat_red_zones.csv",
    "safehabitat_relocation_priority.csv",
    "safehabitat_candidate_site_results.csv",
    "safehabitat_model_performance.csv",
    model_filename
]

for filename in output_files:
    files.download(filename)

print("\n" + "=" * 60)
print("SAFEHABITAT AI EXECUTION COMPLETED")
print("=" * 60)
```
